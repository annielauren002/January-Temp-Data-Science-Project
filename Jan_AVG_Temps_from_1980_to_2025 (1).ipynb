```python
import pandas as pd
import plotly.express as px
```

```python
jan_1980 = pd.read_csv('Jan 1980.csv')
jan_1990 = pd.read_csv('Jan 1990.csv')
jan_2000 = pd.read_csv('Jan 2000.csv')
jan_2010 = pd.read_csv('Jan 2010.csv')
jan_2020 = pd.read_csv('Jan 2020.csv')
jan_2025 = pd.read_csv('Jan 2025.csv')

jan_1980.info()
jan_1990.info()
jan_2000.info()
jan_2010.info()
jan_2020.info()
```

```python
# Calculating average temp from TMAX and TMIN and putting it in TAVG column
# 1980
jan_1980['TAVG'] = jan_1980[['TMAX', 'TMIN']].mean(axis=1)
jan_1980.info()

# 1990
jan_1990['TAVG'] = jan_1990[['TMAX', 'TMIN']].mean(axis=1)
jan_1990.info()

# 2000
jan_2000['TAVG'] = jan_2000[['TMAX', 'TMIN']].mean(axis=1)
jan_2000.info()

# 2010
jan_2010['TAVG'] = jan_2010[['TMAX', 'TMIN']].mean(axis=1)
jan_2010.info()

# 2020
jan_2020['TAVG'] = jan_2020[['TMAX', 'TMIN']].mean(axis=1)
jan_2020.info()

# 2025
jan_2025['TAVG'] = jan_2025[['TMAX', 'TMIN']].mean(axis=1)
jan_2025.info()
```

```python
# concatenating all jan years
jan_df = pd.concat([jan_1980, jan_1990, jan_2000, jan_2010, jan_2020, jan_2025])
jan_df.head()

# chosing columns i want
jan_df = jan_df[['DATE', 'PRCP', 'SNOW', 'TAVG', 'TMAX', 'TMIN']]
jan_df.head()
jan_df.info()

# exporting mega file
jan_df.to_csv('jan_df.csv', index=False)
```

```python
# updating DATE data type
jan_df['DATE'] = pd.to_datetime(jan_df['DATE'], format='mixed')
jan_df.info()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
jan_df['YEAR'] = jan_df['DATE'].dt.year

# Average temp for each January over 50 years
yearly_avg = jan_df.groupby('YEAR')['TAVG'].mean().reset_index()


plt.figure(figsize=(10,6))
sns.lineplot(data=yearly_avg, x='YEAR', y='TAVG', marker='o')

plt.title('Average January Temperature (1980–2020)')
plt.xlabel('Year')
plt.ylabel('Avg Temperature (°F)')
plt.grid(True)
plt.tight_layout()
plt.show()

```

```python
fig = px.line(
    yearly_avg,
    x='YEAR',
    y='TAVG',
    markers=True,
    title='Average January Temperature (1980–2025)'
)

fig.update_layout(
    xaxis_title='Year',
    yaxis_title='Avg Temperature (°F)',
    template='plotly_white'
)

fig.show()

fig.write_html("avg_january_temp_trend.html")
```

```python
jan_df['DAY'] = jan_df['DATE'].dt.day
jan_df['YEAR'] = jan_df['DATE'].dt.year


yearly_avg_jan = jan_df.groupby(['YEAR'])['TAVG'].mean().reset_index()
```

```python
# bar chart showing the average temp vs year
plt.figure(figsize=(10,6))
sns.barplot(data=yearly_avg_jan, x='YEAR', y='TAVG', palette='coolwarm')
plt.title('Average January Temperature (1980–2025)')
plt.xlabel('Year')
plt.ylabel('Avg Temperature (°F)')

# Add value labels on top of bars
for i, row in yearly_avg.iterrows():
    plt.text(i, row['TAVG'] + 0.5, f"{row['TAVG']:.1f}",
             ha='center', fontsize=9)


plt.tight_layout()
plt.show()
```

```python
# interactive bar chart
fig = px.bar(
    yearly_avg,  # or yearly_avg_jan if you're using that
    x='YEAR',
    y='TAVG',
    text='TAVG',  # adds value labels
    color='TAVG',  # for visual appeal like coolwarm
    color_continuous_scale='RdYlBu',  # similar to coolwarm
    title='Average January Temperature (1980–2025)',
)

# adjusting where temp avg is located
fig.update_traces(
    texttemplate='%{text:.1f}',
    textposition='outside'
)

# updating titles
fig.update_layout(
    xaxis_title='Year',
    yaxis_title='Avg Temperature (°F)',
    uniformtext_minsize=8,
    uniformtext_mode='hide',
    template='plotly_white'
)

fig.show()

fig.write_html("jan_bar_plotly.html")
```

```python
# creat box plots
plt.figure(figsize=(10,6))
sns.boxplot(data=jan_df, x='YEAR', y='TAVG', palette='coolwarm')

# Overlay with strip plot for actual data points
sns.stripplot(data=jan_df, x='YEAR', y='TAVG', color='black', alpha=0.4, size=3)

plt.title('Average January Temperature (1980–2025)')
plt.xlabel('Year')
plt.ylabel('Avg Temperature (°F)')

# Add reference line for freezing point
plt.axhline(y=32, color='red', linestyle='--', alpha=0.5, label='Freezing Point')
plt.legend()

plt.tight_layout()
```

```python
# interactive box plot
fig = px.box(
    jan_df,
    x='YEAR',
    y='TAVG',
    color='YEAR',
    points='all',
    title='Average January Temperature (1980–2025)',
    color_discrete_sequence=px.colors.qualitative.Pastel
)

#adding freezing point line
fig.add_hline(
    y=32,
    line_dash='dash',
    line_color='red',
    annotation_text='Freezing Point (32°F)',
    annotation_position='bottom left',
    opacity=0.5
)

#updating titles
fig.update_layout(
    xaxis_title='Year',
    yaxis_title='Avg Temperature (°F)',
    template='plotly_white'
)

fig.show()

fig.write_html("jan_box_plotly.html")
```

```python
# pivot table for heat map
pivot_table = jan_df.pivot_table(index='YEAR', columns='DAY', values='TAVG', aggfunc='mean')
pivot_table

# heat map
plt.figure(figsize=(20, 10))
sns.heatmap(pivot_table, cmap='RdYlBu', annot=True, fmt='.1f', linewidths=0.5)

plt.title('Average January Temperature (1980–2025)')
plt.xlabel('Day of the Month')
plt.ylabel('Year')

plt.tight_layout()
plt.show()

plt.savefig('jan_heatmap.png')
```

```python
import plotly.express as px

# You already have this:
pivot_table = jan_df.pivot_table(index='YEAR', columns='DAY', values='TAVG', aggfunc='mean')

# Create interactive heatmap
fig = px.imshow(
    pivot_table.values,
    labels=dict(x="Day of the Month", y="Year", color="Avg Temp (°F)"),
    x=pivot_table.columns,  # Days (1–31)
    y=pivot_table.index,    # Years (1980–2025)
    color_continuous_scale='RdYlBu',  # similar to Seaborn's coolwarm
    aspect='auto',
    title='Interactive Heatmap: Average January Temperature (1980–2025)'
)

fig.update_layout(
    template='plotly_white',
    xaxis_nticks=31,
    yaxis_autorange='reversed'
)

fig.show()

fig.write_html("jan_heatmap_plotly.html")
```
